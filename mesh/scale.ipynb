```python
def scale_ply(input_file, output_file, scale_factor=0.125):
    with open(input_file, 'r') as f:
        lines = f.readlines()

    # Identify header boundaries
    header_end = 0
    vertex_count = 0
    face_count = 0

    for i, line in enumerate(lines):
        if line.startswith('element vertex'):
            # Example: "element vertex 126411"
            vertex_count = int(line.split()[2])
        elif line.startswith('element face'):
            # Example: "element face 252818"
            face_count = int(line.split()[2])
        elif line.strip() == 'end_header':
            header_end = i
            break

    # The vertex data should start right after the header
    vertex_start = header_end + 1
    vertex_end = vertex_start + vertex_count

    # Process vertex lines
    scaled_vertices = []
    for v_line in lines[vertex_start:vertex_end]:
        parts = v_line.split()
        # Assuming the first three parts are x, y, z
        x, y, z = float(parts[0]), float(parts[1]), float(parts[2])
        # Scale coordinates
        x_scaled = x * scale_factor
        y_scaled = y * scale_factor
        z_scaled = z * scale_factor
        
        # Reconstruct the line with scaled coordinates and the original color values
        new_line = f"{x_scaled} {y_scaled} {z_scaled} {' '.join(parts[3:])}\n"
        scaled_vertices.append(new_line)

    # The face data starts after the vertex data
    face_start = vertex_end
    face_end = face_start + face_count
    face_data = lines[face_start:face_end]

    # Write the new file
    with open(output_file, 'w') as out:
        # Write the header
        out.writelines(lines[:vertex_start])
        # Write scaled vertex lines
        out.writelines(scaled_vertices)
        # Write face lines
        out.writelines(face_data)
```
